# Visualization and Analysis of Embeddings

This notebook starts after embeddings already exist in AnnData. In real
workflows, create those matrices with `BioEmbedder.embed(...)`; the plotting
and metric utilities below only consume `.obsm` keys.


```python
import warnings
warnings.filterwarnings("ignore")

import anndata as ad
import numpy as np
import pandas as pd
import embpy.pl as pl
import embpy.tl as tl
```

## 1. Minimal synthetic AnnData for plotting examples


```python
rng = np.random.default_rng(42)
n = 50
adata = ad.AnnData(
    X=np.zeros((n, 1), dtype=np.float32),
    obs=pd.DataFrame(
        {
            "perturbation": [f"gene_{i % 10}" for i in range(n)],
            "mechanism": np.where(np.arange(n) % 2 == 0, "repair", "signaling"),
        },
    ),
)
adata.obs_names = [f"obs_{i}" for i in range(n)]
adata.obsm["X_demo_a"] = rng.normal(size=(n, 16)).astype(np.float32)
adata.obsm["X_demo_b"] = rng.normal(size=(n, 16)).astype(np.float32)
adata
```

## 2. Dimensionality reduction and quick plots


```python
tl.compute_umap(adata, obsm_key="X_demo_a", output_key="X_umap_demo_a", n_neighbors=8)
fig = pl.plot_embedding_space(
    adata,
    obsm_key="X_demo_a",
    basis="X_umap_demo_a",
    color="mechanism",
    title="Demo embedding space",
)
fig
```

## 3. Similarity and cross-embedding diagnostics


```python
fig = pl.plot_similarity_heatmap(
    adata,
    obsm_key="X_demo_a",
    labels=list(adata.obs["perturbation"]),
    title="Cosine similarity",
)
fig
```

```python
fig = pl.cross_model_similarity(
    adata,
    obsm_keys=["X_demo_a", "X_demo_b"],
    labels=["demo A", "demo B"],
)
fig
```
