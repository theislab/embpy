# Tutorial 9: Morphology and Perturbation Embeddings

Morphology-derived perturbation embeddings are generated through
`BioEmbedder.embed(...)` with `entity_type="perturbation"`. The source of
truth is stored in `.uns["perturbations"][key]`; downstream models can
explicitly project that entity-aligned payload into `.obsm` when they need
one vector per observation.


```python
from embpy import BioEmbedder

RUN_EMBEDDING = False  # Set True when you are ready to run model inference.
embedder = BioEmbedder(device="auto", organism="human")
print(f"device={embedder.device}")
```

## 1. HPA/SubCell perturbation embeddings


```python
genes = ["TP53", "BRCA1", "EGFR"]

if RUN_EMBEDDING:
    hpa_adata = embedder.embed(
        genes,
        entity_type="perturbation",
        model="subcell_mae_rybg",
        output="anndata",
        attach_to="uns",
        key="X_pert_subcell_mae_rybg",
        morphology_dataset="hpa",
        morphology_source="subcell",
        morphology_local_dir="data/embeddings/morphology_cache/hpa",
        max_images=5,
        aggregation="mean",
        morphology_workers=8,
    )
    payload = hpa_adata.uns["perturbations"]["X_pert_subcell_mae_rybg"]
    print(payload["n_entities"], payload["n_dims"])
```

## 2. JUMP/SubCell perturbation embeddings


```python
if RUN_EMBEDDING:
    jump_adata = embedder.embed(
        genes,
        entity_type="perturbation",
        model="subcell_mae_rybg",
        output="anndata",
        attach_to="uns",
        key="X_pert_jump_subcell_mae_rybg",
        morphology_dataset="jump",
        morphology_source="subcell",
        morphology_local_dir="data/embeddings/morphology_cache/jump",
        max_images=5,
        aggregation="mean",
        morphology_workers=8,
    )
    print(jump_adata.uns["perturbations"].keys())
```

## 3. Persist perturbation payloads as Zarr or CSV tables


```python
if RUN_EMBEDDING:
    embedder.embed(
        genes,
        entity_type="perturbation",
        model="subcell_mae_rybg",
        output="table",
        path="perturbation_subcell_embeddings.zarr",
        fmt="zarr",
        morphology_dataset="hpa",
        morphology_source="subcell",
        morphology_local_dir="data/embeddings/morphology_cache/hpa",
        max_images=5,
    )
```
