# Tutorial 13: DepMap Cancer Dependency Analysis

DepMap matrices are phenotypic measurements. When the analysis needs gene,
molecule, or context embeddings, those embeddings are generated with
`BioEmbedder.embed(...)` and then joined back to the DepMap table.


```python
from embpy import BioEmbedder
import embpy
import numpy as np
import pandas as pd

RUN_EMBEDDING = False
DATA_DIR = "data/datasets/depmap"
embedder = BioEmbedder(device="auto", organism="human")
```

## 1. Load a DepMap phenotype matrix


```python
# Adjust DATA_DIR to your local DepMap cache.
if False:
    adata = embpy.pp.load_depmap("CRISPRGeneEffect", data_dir=DATA_DIR)
    print(adata)
```

## 2. Embed DepMap genes through the unified API


```python
depmap_genes = ["TP53", "KRAS", "EGFR", "MYC", "BRAF"]

if RUN_EMBEDDING:
    depmap_gene_embeddings = embedder.embed(
        depmap_genes,
        entity_type="gene",
        id_type="symbol",
        model="esm2_650M",
        output="anndata",
        key="X_depmap_gene_esm2_650M",
    )
    print(depmap_gene_embeddings.varm["X_depmap_gene_esm2_650M"].shape)
```

## 3. Pair phenotypes with embeddings


```python
if RUN_EMBEDDING:
    # Example shape: rows are genes, columns are embedding dimensions.
    X_gene = depmap_gene_embeddings.varm["X_depmap_gene_esm2_650M"]
    gene_table = pd.DataFrame(
        X_gene,
        index=depmap_gene_embeddings.var_names,
        columns=[f"esm2_{i}" for i in range(X_gene.shape[1])],
    )
    print(gene_table.head())
```

## 4. Save reusable gene embeddings


```python
if RUN_EMBEDDING:
    embedder.embed(
        depmap_genes,
        entity_type="gene",
        id_type="symbol",
        model="esm2_650M",
        output="table",
        path="depmap_gene_embeddings.csv",
        fmt="csv",
    )
```
