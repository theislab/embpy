# Tutorial 4: Structure-Aware Protein Embeddings

Boltz-2 and sequence protein models are both accessed through
`BioEmbedder.embed(...)`. Optional structure dependencies are handled by the
model environment; the tutorial keeps the public API identical.


```python
from embpy import BioEmbedder

RUN_EMBEDDING = False  # Set True when you are ready to run model inference.
embedder = BioEmbedder(device="auto", organism="human")
print(f"device={embedder.device}")
```

## 1. Structure-aware embeddings from gene symbols


```python
proteins = ["TP53", "BRCA1", "EGFR"]

if RUN_EMBEDDING:
    boltz_payload = embedder.embed(
        proteins,
        entity_type="protein",
        id_type="symbol",
        model="boltz2",
        output="payload",
        key="X_boltz2",
        pooling_strategy="mean",
    )
    print(boltz_payload["n_entities"], boltz_payload["n_dims"])
```

## 2. Pair sequence and structure models in one comparison object


```python
if RUN_EMBEDDING:
    comparison = embedder.embed(
        proteins,
        entity_type="protein",
        id_type="symbol",
        model=["esm2_650M", "boltz2"],
        output="anndata",
        harmonize_dim=128,
    )
    print(list(comparison.varm.keys()))
```

## 3. Pairwise or combined Boltz-2 outputs


```python
if RUN_EMBEDDING:
    pairwise = embedder.embed(
        proteins,
        entity_type="protein",
        id_type="symbol",
        model="boltz2_pairwise",
        output="payload",
        key="X_boltz2_pairwise",
    )
    both = embedder.embed(
        proteins,
        entity_type="protein",
        id_type="symbol",
        model="boltz2_both",
        output="payload",
        key="X_boltz2_both",
    )
    print(pairwise["model"], both["model"])
```
