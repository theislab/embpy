# Tutorial 5: Molecule Embeddings

Molecules use the same public embedding API as genes and proteins. Pass
names or SMILES, choose a molecule model, and decide whether the artifact
should be AnnData, a payload, CSV, or Zarr.


```python
from embpy import BioEmbedder

RUN_EMBEDDING = False
embedder = BioEmbedder(device="auto")
```

## 1. SMILES to AnnData


```python
smiles = [
    "CC(=O)Oc1ccccc1C(O)=O",      # aspirin
    "Cn1c(=O)c2c(ncn2C)n(C)c1=O", # caffeine
    "CC(C)Cc1ccc(cc1)C(C)C(O)=O", # ibuprofen
]

if RUN_EMBEDDING:
    mol_adata = embedder.embed(
        smiles,
        entity_type="molecule",
        id_type="smiles",
        model="chemberta2MTR",
        output="anndata",
        key="X_chemberta2MTR",
    )
    print(mol_adata.obsm["X_chemberta2MTR"].shape)
    print(mol_adata.obs.head())
```

## 2. Names resolve to canonical SMILES


```python
names = ["aspirin", "caffeine", "ibuprofen"]

if RUN_EMBEDDING:
    named = embedder.embed(
        names,
        entity_type="molecule",
        id_type="name",
        model="morgan_fp",
        output="payload",
        key="X_morgan_fp",
    )
    print(named["id_scheme"])
    print(named["alias_schemes"])
```

## 3. Compare several molecule featurizers


```python
if RUN_EMBEDDING:
    mol_multi = embedder.embed(
        smiles,
        entity_type="molecule",
        id_type="smiles",
        model=["chemberta2MTR", "morgan_fp", "maccs_fp"],
        output="anndata",
        harmonize_dim=64,
    )
    print(list(mol_multi.obsm.keys()))
```

## 4. Save scverse-friendly outputs


```python
if RUN_EMBEDDING:
    embedder.embed(
        smiles,
        entity_type="molecule",
        id_type="smiles",
        model="morgan_fp",
        output="table",
        path="molecule_embeddings_morgan.zarr",
        fmt="zarr",
    )
```
